```python
letters = ['a', 'b', 'c', 'd']
print (letters[int(int('3'*2)//11)] )
# '3'*2 gives '33', then int('33') is 33, then 33//11 is 3, then letter[3] is 'd'
print (letters[-1])
# [-1] means the last letter, so it's 'd'
print(letters[:2])
# [:2] means the first two letters
```
