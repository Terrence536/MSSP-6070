## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
sum(salaries)
len(salaries)
mean = sum(salaries)/len(salaries)
mean
```

```python
import statistics
statistics.mean(salaries)
```

### median

```python
sorted_salaries = sorted(salaries)
median = (sorted_salaries[49] + sorted_salaries[50]) / 2
median
```

```python
import statistics
statistics.median(salaries)
```

### mode

```python
import statistics
statistics.mode(salaries)
```

### sample variance
Remember to use Bessel's correction.

```python

sample_variance = sum((x - mean) ** 2 for x in salaries) / (len(salaries) - 1)

sample_variance
```

```python
import statistics
statistics.variance(salaries)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_standard_deviation = sample_variance ** 0.5
sample_standard_deviation
```

```python
import statistics
statistics.stdev(salaries)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_salary = min(salaries)
max_salary = max(salaries)

min_max_scaled = [
    (x - min_salary) / (max_salary - min_salary)
    for x in salaries
]

min_max_scaled[:5]
```

### standardizing

```python
from statistics import mean, stdev

mean_salary = mean(salaries)
std_salary = stdev(salaries)

standardized = [
    (x - mean_salary) / std_salary
    for x in salaries
]

standardized[:5]
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import numpy as np

np.cov(min_max_scaled, standardized)
```

### Pearson correlation coefficient ($\rho$)

```python
np.corrcoef(min_max_scaled, standardized)
```
