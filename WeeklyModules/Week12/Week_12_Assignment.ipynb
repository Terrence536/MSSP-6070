```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

df = pd.read_csv("data/Student Performance Data.csv")

df.head()
```

```python
print(df.shape)
print(df.columns.tolist())
print(df.isnull().sum())
```

```python
male_df = df[df["gender"] == "male"]

male_df["test preparation course"].value_counts()
```

```python
male_df.groupby("test preparation course")[
    ["math score", "reading score", "writing score", "total score"]
].mean()
```

```python
(df["total score"] == df[
    ["math score", "reading score", "writing score"]
].sum(axis=1)).all()
```

```python
male_df = df[df["gender"] == "male"].copy()

male_df["average score"] = male_df["total score"] / 3

completed = male_df[
    male_df["test preparation course"] == "completed"
]["average score"]

none = male_df[
    male_df["test preparation course"] == "none"
]["average score"]

print("Completed mean:", completed.mean())
print("None mean:", none.mean())

t_stat, p_value = stats.ttest_ind(
    completed, none, equal_var=False)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

An independent samples t-test (Welch's t-test) was conducted to examine whether there was a statistically significant difference in average scores between male students who completed the test preparation course and those who did not.

The results show that male students who completed the course had a higher average score (M = 70.78) than those who did not (M = 63.04), with a mean difference of 7.74 points. This difference was statistically significant (t = 6.18, p < 0.001).

Therefore, we reject the null hypothesis and conclude that there is a statistically significant difference in average scores between the two groups.
