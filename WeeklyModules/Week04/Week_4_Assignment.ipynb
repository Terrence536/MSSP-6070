## d

```python
df["parsed_place"].value_counts().head(20)
```

```python
ring_locations = ["Bolivia", "Chile", "Ecuador", "Peru", "Costa Rica",
    "Guatemala", "Mexico", "Japan", "Philippines",
    "Indonesia", "New Zealand", "Canada", "Fiji",
    "Alaska", "Washington", "California", "Russia",
    "Taiwan", "Tonga", "Kermadec Islands"]
```

```python
df.loc[df["parsed_place"].str.contains("Antarctic", case=False, na=False),
    "parsed_place"].value_counts()
```

```python
df["ring_of_fire"] = (df["parsed_place"].isin(ring_locations)
    | df["parsed_place"].str.contains("Antarctic", case=False, na=False))

df[["parsed_place", "ring_of_fire"]].head()
```

```python
df.loc[df["parsed_place"].isin(["Mexico", "New Mexico"]),
    ["parsed_place", "ring_of_fire"]].drop_duplicates()
```

## e

```python
df["ring_of_fire"].value_counts()
```

There are 7188 earthquakes in the Ring of Fire locations and 2144 out of the locations.
